This code will merge all Excel files into a single excel sheet

```python
import os
import pandas as pd

# ============================================================
# STEP 1: Input and output paths
# ============================================================

INPUT_DIR = r"C:/Users/Juvair/Desktop/Job Application CV's/BUET_RA/Dataset/Air Pollution Dataset/CAMS_CleanDataset/Renamed"

OUTPUT_FILE = os.path.join(
    INPUT_DIR,
    "CAMS_All_Stations.xlsx"
)
```

```python
# ============================================================
# STEP 2: Find all Excel files
# ============================================================

csv_files = [
    os.path.join(INPUT_DIR, file)
    for file in os.listdir(INPUT_DIR)
    if file.endswith((".csv"))
    and file != "CAMS_All_Stations.xlsx"
]
print(f"Found {len(csv_files)} CSV files.")
```

```python
# ============================================================
# STEP 3: Read and merge all files
# ============================================================

all_data = []

for file_path in csv_files:

    file_name = os.path.basename(file_path)

    print(f"Processing: {file_name}")

    # Read CSV
    df = pd.read_csv(file_path)

    # Station name from filename
    station_name = os.path.splitext(file_name)[0]

    # ========================================================
    # Add CAMS_Station immediately after CalendarDate
    # ========================================================

    if "CalendarDate" in df.columns:

        date_position = df.columns.get_loc("CalendarDate")

        df.insert(
            date_position + 1,
            "CAMS_Station",
            station_name
        )

    else:
        print(
            f"Warning: 'CalendarDate' not found in {file_name}"
        )

        df.insert(
            0,
            "CAMS_Station",
            station_name
        )

    # Add to master list
    all_data.append(df)
```

```python
# ========================================================
# STEP 5: Save merged Excel file
# ========================================================

if all_data:

    merged_data = pd.concat(
        all_data,
        ignore_index=True
    )
    merged_data.to_excel(
        OUTPUT_FILE,
        index=False
    )

    print("\n==============================================")
    print("DONE!")
    print("==============================================")

    print(f"\nOutput file:")
    print(OUTPUT_FILE)

    print(f"\nTotal rows: {len(merged_data):,}")
    print(f"Total columns: {len(merged_data.columns)}")

    print("\nCAMS stations:")
    print(merged_data["CAMS_Station"].unique())

    print("\nPreview:")
    print(merged_data.head())

else:

    print("\nNo data found to merge.")
```
